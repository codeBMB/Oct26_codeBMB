# 3. Write sequence-analysis functions

A function packages a calculation so that you can reuse it. We will write a function that calculates the percentage of guanine and cytosine bases in a DNA sequence.

Before applying a function to a whole column, test it on one sequence. Notice that biological data can contain lowercase letters, so the function should standardize the sequence first.


```python
def gc_content(seq):
    seq = seq.upper()
    return 100 * (seq.count("G") + seq.count("C")) / len(seq)

example_dna = "ATGCGC"
print(gc_content(example_dna))
```

```python
# Exercise 3 — solution
def count_motif(seq, motif):
    seq = seq.upper()
    motif = motif.upper()
    return seq.count(motif)

print(count_motif("ATGCGCATG", "ATG"))
```

# 4. Add calculated columns and summarize

Now apply your functions to the DataFrame. Protein sequences do not have meaningful DNA GC content, so calculate GC content only for rows whose `molecule_type` is `"DNA"`. For every sequence, calculate its length.

Then compare sequence lengths by molecule type.


```python
# Exercise 4 — solution
df["length"] = df["sequence"].str.len()
df["gc_content"] = df.apply(
    lambda row: gc_content(row["sequence"]) if row["molecule_type"] == "DNA" else pd.NA,
    axis=1
)
df["has_motif_ATG"] = df["sequence"].str.upper().str.contains("ATG")

print(df[["sample_id", "molecule_type", "length", "gc_content", "has_motif_ATG"]])
print("\nMean length by molecule type:")
print(df.groupby("molecule_type")["length"].mean())
```

# 5. Visualize the sequence properties

A plot can reveal patterns that are difficult to see in a table. Create a scatter plot of sequence length versus GC content for the DNA sequences. Label the axes and give the plot a descriptive title.

Because protein rows have missing GC content, filter them out before plotting.


```python
import matplotlib.pyplot as plt

dna_df = df[df["molecule_type"] == "DNA"].copy()

# Example plotting pattern:
# plt.scatter(dna_df["length"], dna_df["gc_content"])
# plt.xlabel("Sequence length (nucleotides)")
# plt.ylabel("GC content (%)")
# plt.title("GC content and sequence length")
# plt.show()
```

```python
# Exercise 5 — solution
plt.figure(figsize=(7, 4))
plt.scatter(dna_df["length"], dna_df["gc_content"], s=80)

for _, row in dna_df.iterrows():
    plt.annotate(row["sample_id"], (row["length"], row["gc_content"]))

plt.xlabel("Sequence length (nucleotides)")
plt.ylabel("GC content (%)")
plt.title("GC content and sequence length")
plt.show()
```
