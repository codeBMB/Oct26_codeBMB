### 💡 How to Hide This Solution

If you're like me, you want your users to make an effort on their own before they just copy and paste some code. It's pretty simple. Here's an example of the Python that you'll need in Colab. The next coding cell contains the same example.

To make this notebook clean or ready for students, you can hide the solution cell's code:

1. **Add Form Metadata**: Put `# @title Solution { display-mode: "form" }` at the very top of the code cell.
2. **Collapse Code**: Click the **three vertical dots** on the top right of the cell, then select **Form** -> **Hide code**.
3. **Clear Outputs**: To hide the printed output, click the `X` next to the output or select **Clear output** from the cell menu.

## An example of hiding code
Here is a simple problem that you might give your students.

*Create a for loop that will create a list of enzyme velocities based on Km = 15 $\mu$M, Vmax = 100 nM/min, and substrate concentrations of 1, 2, 3, 5, 10, 20, 40, 60, and 100 $\mu$M.*


```python
# @title Solution { display-mode: "form" }

list = [1, 2, 3, 5, 10, 20, 40, 60, 100] # micromolar substrate concentrations
Km = 15
Vmax = 100
velocities = []
for item in list:
  v = item*Vmax/(Km+item)
  velocities.append(v)
print(velocities)
```
