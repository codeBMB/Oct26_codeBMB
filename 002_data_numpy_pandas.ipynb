A **List** is Python's built-in way to store a collection of items. They are extremely flexible and can hold different types of data (integers, strings, even other lists) all in one place.

A list is best for general-purpose storage of small collections of data or simple sequences.

Characteristics of a list in Python:
*  Easy to create and modify.
*  Can store mixed data types.
*  Slower for mathematical operations on large datasets.

````python
# A list of mixed types
fruits = ["apple", 42, True, "banana"]

# A list of integers
scores = [85, 90, 78, 92]

print(fruits)

print(scores)
````

A NumPy **array** (or ndarray) is the foundation of the NumPy library. Unlike lists, arrays are designed specifically for numerical computing.

Arrays are best for fast mathematical operations, linear algebra, and multi-dimensional data.

Characteristics:
*  All elements must be of the **same** type (e.g., all floats).
*  You can perform operations on the entire array at once (e.g., array * 2 multiplies every element by 2).
*  Uses significantly less memory than lists for large datasets.

````python
# a list of substrate concentrations
substrate_list = [1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0] #micromolar


# an array of substrate concentations
substrate_concs = np.array([1.0, 2.0, 3.0, 4.0, 6.0, 8.0, 10.0, 15.0, 20.0, 30.0, 50.0, 75.0, 100.0]) #micromolar

print(substrate_list)
print(substrate_concs)
````

`milli_list = [x / 1000 for x in substrate_list]`

*  **The Operation (x / 1000):** This is the Instruction. This is what you want to do to the number you are currently holding.You are telling Python: "Take whatever x is, and divide it by 1000."

*  **The Result ([...]):** The brackets tell Python to collect all those results into a new list. Without the brackets, you would just have a bunch of numbers floating around; with the brackets, you have a clean, new list.

## Generating mock datasets

For a first-time user, the np.____ and pd.____ syntax can feel like "magic" until you understand the hierarchy. It follows a very logical pattern: Library → Tool → Action. We've seen object oriented calls as well where the pattern is Object → Tool/Action (e.g. `kinetics.to_csv()`)

### Summary Table: How to setup the function

| Syntax | What it means | Where the tool lives | Example from your code |
| :--- | :--- | :--- | :--- |
| **`np.____()`** | "Go to the **NumPy** library and perform this general math/random task." | Inside the **Library** | `np.random.uniform()` |
| **`df.____()`** | "Take the **DataFrame** I just made and perform this specific action on it." | Inside the **Object** | `df.to_csv()` |


This code below demonstrates how to use NumPy and Pandas to generate a realistic, synthetic dataset of plant growth parameters for student practice. NumPy provides the core engine for generating various types of random data using functions like `np.random.choice`, `np.random.uniform`, `np.random.normal`, and `np.random.exponential`, while also performing vectorized math to calculate dependent variables like height and leaf count. Finally, Pandas organizes these individual arrays into a structured `pd.DataFrame`, providing a tabular format where missing values can be injected using `.loc` and `np.nan` to simulate real-world data cleaning challenges.

````python
# making data sets for students testing

np.random.seed()  # For reproducibility
n = 200 # number of rows

# Dependent variable
species      = np.random.choice(["A", "B", "C"], # categorical variable
                                size=n, # number of objects
                                p=[0.4, 0.35, 0.25]) # frequency of A,B,C
light_hours  = np.random.uniform(4, 14, # uniform distribution of random numbers between 4 and 14 
                                 size=n).round(1) #rounded to 1 decimal
water_ml     = np.random.normal(250, 50, # normal distribution, mean of 250 and std_dev of 50
                                size=n).clip(80, 450).round(0) # floor value of 80, ceiling of 450
fertilizer_g = np.random.exponential(scale=5, # decay curve, 5 indicates steepness
                                     size=n).clip(0, 25).round(1) # floor value of 0, ceiling of 25
temperature_c = np.random.normal(22, 4, # normal distribution, mean of 22 and std_dev of 4
                                 size=n).round(1)

# Independent variables
# Height depends on light, water, fertiliser, temperature, and species
species_bonus = {"A": 0, "B": 8, "C": -5}
height_cm = (
    10
    + 2.1  * light_hours
    + 0.05 * water_ml
    + 1.9  * fertilizer_g
    - 0.3  * (temperature_c - 22) ** 2   # peaks near 22 °C
    + np.array([species_bonus[s] for s in species]) # list for loop
    + np.random.normal(0, 4, size=n) # add in random noise, normal distribution
).clip(5, None).round(1) # height has to be between 5 and infinity

# assemble the data frame
df = pd.DataFrame({
    "plant_id":     range(1, n + 1), # create a sequence of number from 1 to n + 1
    "species":      species,
    "light_hours":  light_hours,
    "water_ml":     water_ml,
    "fertilizer_g": fertilizer_g,
    "temperature_c":temperature_c,
    "height_cm":    height_cm
})

# Inject missing values into water_ml to practice cleaning
missing_idx_w = np.random.choice(df.index, size=13, replace=False)
df.loc[missing_idx_w, "water_ml"]     = np.nan

print(f"Dataset created: {df.shape[0]} rows × {df.shape[1]} columns")
````


### Descriptive Statistics

Descriptive statistics summarise the central tendency, spread, and shape of each variable. They're the fastest way to get a quantitative feel for your data.

Notice the command structure shown in the **Pandas/NumPy** column of the table below. Pandas uses the dataframe name, while NumPy uses the package reference (`np`) followed by the method.

### Key statistics to know:

| Method | Meaning | Pandas | NumPy|
|---|---|---|---|
| **Mean** | Average value | `df.mean()` |  `np.mean()` |
| **Median** | Middle value (50th percentile) | `df.median()` | `np.median()` |
| **Std dev** | Spread around the mean | `df.std()` | `np.std()`  |
| **Min / Max** | Extreme values | `df.min()` or `df.max()` |
| **Percentiles** | Value at the *p*th rank | `df.quantile(p)` | `np.percentile()` |
| **Skewness** | Asymmetry of distribution | `df.skew()` |
| **Kurtosis** | Heaviness of tails | `df.kurt()` |
|**Describe** | Basic statistics in one command | `df.describe()` |
